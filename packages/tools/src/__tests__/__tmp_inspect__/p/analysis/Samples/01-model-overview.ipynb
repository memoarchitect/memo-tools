# Model overview

Visual summary of model size, semantic content, and diagnostics.

This sample uses only standard SysML v2 concepts and reads the model through `memo_model.py`, compiled by the project's selected compiler.

```python
import sys
from pathlib import Path
from collections import Counter, defaultdict

# memo_model.py sits in the project's analysis/ folder, above this notebook.
sys.path.insert(0, str(next(d for d in (Path.cwd(), *Path.cwd().parents) if (d / 'memo_model.py').is_file())))
import memo_model

# Compiled by the project's selected compiler (toolchain.compiler), as every
# MEMO command is: change the setting and this notebook follows.
model = memo_model.load()
diagnostics = model.diagnostics
SYSML_FILES = model.documents
print(f'Loaded {len(SYSML_FILES)} SysML files, compiled by {model.compiler} {model.compiler_version or ""}')
```

```python
from IPython.display import HTML, display

semantic_types = [
    'Package', 'PartDefinition', 'PartUsage', 'RequirementDefinition',
    'RequirementUsage', 'ActionDefinition', 'ActionUsage', 'PortDefinition',
    'PortUsage', 'ConnectionDefinition', 'ConnectionUsage', 'StateDefinition',
    'StateUsage', 'AttributeDefinition', 'AttributeUsage'
]
counts = {}
for type_name in semantic_types:
    counts[type_name] = len(model.elements(type_name))

largest = max(counts.values(), default=1)
rows = ''.join(
    f'<tr><td>{name}</td><td>{count:,}</td><td><div style="background:#2d8d70;height:12px;width:{max(2, count * 320 / largest):.0f}px"></div></td></tr>'
    for name, count in sorted(counts.items(), key=lambda item: item[1], reverse=True)
)
display(HTML(f'<h3>Semantic inventory</h3><table><tr><th>Element type</th><th>Count</th><th>Relative size</th></tr>{rows}</table>'))

{
    'SysML files': len(SYSML_FILES),
    'Errors': len(diagnostics.errors),
    'Warnings': len(diagnostics.warnings),
    'Information': len(diagnostics.infos),
}
```
